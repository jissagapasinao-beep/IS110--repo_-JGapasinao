```python
import numpy as np  # Import NumPy library for array operations
```

```python
# Create a nested Python list with values from 1 to 9
list_data = [[1, 2, 3],
             [4, 5, 6],
             [7, 8, 9]]

# Convert list into NumPy ndarray
A = np.array(list_data)

print("Array A:\n", A)  # Display the array

# Using ndarray property
print("\nShape of A:", A.shape)
# .shape shows dimensions (rows, columns)

# Using ndarray method
print("Maximum value using method:", A.max())
# .max() returns the largest value

# Using NumPy function
print("Maximum value using np.max():", np.max(A))
# np.max() also returns the largest value
```

```python
# Create a 5x5 array filled with ones (integers)
B = np.ones((5, 5), dtype=int)

print("Array B:\n", B)

# Check shape
print("\nShape:", B.shape)
# Returns (5,5)

# Check size
print("Size:", B.size)
# Total number of elements (25)

# Check dimensions
print("Number of dimensions:", B.ndim)
# 2 means it's a 2D array
```

```python
# Create array from -5 to +5
C = np.arange(-5, 6)
# 6 is excluded, so values go from -5 to 5

# Multiply each value by 2
C = C * 2

print("Resulting array:\n", C)
```
